# Search-based ST-GCS with per-region cubic Bezier trajectories

This notebook demonstrates the port described in `AGENT/search-based-bezier-implementation-plan.md`:
raising per-GCS-vertex trajectories in this repo's **search-based** ST-GCS solver
(`stgcs/bfs/best_first_search.py`'s `SearchAlgorithm`, `stgcs/gcs_solver.py`'s
`solve_convex_restriction`) from `order=2` (today's piecewise-linear segment) to
`order=4` (one cubic Bezier segment per GCS vertex, per `AGENT/AGENT.md` F1-F8).
Part 8 then stress-tests the whole pipeline with 10 agents planning sequentially
into a shared, initially-empty space. Every figure is interactive Plotly (rotate
the 3D space-time plots, hover for values, toggle traces via the legend).

Reference/companion material:
- `AGENT/AGENT.md` -- the geometric theory (F1-F8) this port implements.
- `AGENT/porting-bezier-to-search-based-stgcs.md` -- the porting plan, now resolved
  (velocity constraint is the L2/SOC form F8 specifies; dominance-check work is
  descoped for this pass -- search runs with a plain duration-based dominance check).
- `notebooks/ecd_bezier_visualization.ipynb` -- single-region ECD walkthrough this
  notebook borrows its visual style (and, for Part 8, its own Part 9-12 stress-test
  precedent) from, adapted to this repo's actual `stgcs.graph`/`stgcs.gcs_solver`/
  `stgcs.ecd` types and the search-based solver (rather than the
  relaxation+rounding-only reference it was written against).

What changed in `stgcs/` to make this possible (see `AGENT/search-based-bezier-implementation-plan.md`
for the full file/line-level plan):
- `stgcs/bezier.py`, `stgcs/hulls.py`, `stgcs/spline_trajectory.py` -- new, standalone
  Bezier geometry (De Casteljau eval/split/time-inversion, Minkowski inflation, the
  `STSplineTrajectory` decode type).
- `stgcs/graph.py` -- `STGCS` takes an `order` parameter; `order>2` gets per-span
  monotone-time, SOC velocity (F8), and C0/C1 edge-continuity (F3) constraints,
  alongside the unchanged `order==2` path.
- `stgcs/gcs_solver.py`, `stgcs/bfs/best_first_search.py` -- decode branches on
  `order` to build an `STSplineTrajectory` instead of `STTrajectory`.
- `stgcs/ecd.py` -- `reserve_spline`/`generate_all_ECD_pairs_spline` reserve a
  Bezier trajectory's Minkowski-inflated per-segment control-point hulls, reusing
  the exact same `slice`/`update_edge` graph-splitting machinery `order==2` uses.

```python
import os
os.environ.setdefault("MOSEKLM_LICENSE_FILE", os.path.expanduser("~/mosek/mosek.lic"))

import time
import numpy as np
import plotly.graph_objects as go
import plotly.colors as pc
from plotly.subplots import make_subplots
from scipy.spatial import ConvexHull

from stgcs.graph import STGCS
from stgcs.interval import Interval
from stgcs.geometry_utils import make_hpolytope, time_extruded
from stgcs.st_planner import MPQuery, SearchPlanner, MICPPlanner, STPlanStatus
from stgcs.bfs.dominance_check import AStarDominanceCheck
from stgcs.bfs.heuristics import MotionOnlyHeuristic
from stgcs.trajectory import STTrajectory
from stgcs.spline_trajectory import STSplineTrajectory
import stgcs.ecd as ecd


def rect(x0, y0, x1, y1):
    return np.array([[x0, y0], [x1, y0], [x1, y1], [x0, y1]])


def run_search(g, start, goal, t_start=0.0, vlimit=1.0, eps=1.0):
    """order-agnostic: SearchPlanner with a plain duration-based dominance check
    (AStarDominanceCheck) and the h_mot heuristic -- neither needs any Bezier-aware
    changes (AGENT/search-based-bezier-implementation-plan.md SS3.1/SS3.2)."""
    q = MPQuery(start=np.array(start, dtype=float), goal=np.array(goal, dtype=float),
                t_start=t_start, is_stay=False, vlimit=vlimit)
    planner = SearchPlanner(dc_list=[AStarDominanceCheck()], heur=MotionOnlyHeuristic(g), eps=eps)
    return planner.plan(g, q)


def sample_curve(sol, n=150):
    ts = np.linspace(sol.x0[-1], sol.xT[-1], n)
    pts = np.array([sol.lerp(t) for t in ts])
    return ts, pts


# --- shared Plotly helpers -----------------------------------------------------
# Same interactive-trace toolkit as notebooks/ecd_bezier_visualization.ipynb, so
# every figure below (not just this notebook's Part 8) is Plotly, not matplotlib:
# rotate/pan/zoom the 3D space-time hulls freely, hover any trace for its
# name/coordinates, toggle traces via the legend.
T10 = pc.qualitative.T10  # same hues/ordering as matplotlib's tab10, used elsewhere in stgcs


def hex_to_rgba(hex_color, alpha):
    hex_color = hex_color.lstrip("#")
    r, g, b = (int(hex_color[i:i + 2], 16) for i in (0, 2, 4))
    return f"rgba({r},{g},{b},{alpha})"


def color_i(i, alpha=1.0):
    return hex_to_rgba(T10[i % len(T10)], alpha)


def hull_polygon(xy):
    hull = ConvexHull(xy)
    pts = xy[hull.vertices]
    return np.vstack([pts, pts[:1]])


def filled_polygon_trace(xy, name, fillcolor, line_color="black", showlegend=True):
    pts = hull_polygon(xy)
    return go.Scatter(x=pts[:, 0], y=pts[:, 1], mode="lines", fill="toself",
                       line=dict(color=line_color, width=1), fillcolor=fillcolor,
                       name=name, showlegend=showlegend, hovertemplate=f"{name}<extra></extra>")


def polygon_trace(xy, name, i, alpha=0.45, line_color="black", showlegend=True):
    return filled_polygon_trace(xy, name, color_i(i, alpha), line_color, showlegend)


def order_points_on_plane(pts):
    """Order near-coplanar 3D points into a boundary loop by angle around their
    centroid in the best-fit plane (SVD normal) -- used only for the rare
    degenerate/near-flat hull `ConvexHull` can't triangulate."""
    centroid = pts.mean(axis=0)
    centered = pts - centroid
    _, _, vt = np.linalg.svd(centered)
    normal = vt[-1]
    arbitrary = np.array([1.0, 0.0, 0.0]) if abs(normal[0]) < 0.9 else np.array([0.0, 1.0, 0.0])
    u = np.cross(normal, arbitrary); u /= np.linalg.norm(u)
    v = np.cross(normal, u)
    angles = np.arctan2(centered @ v, centered @ u)
    return pts[np.argsort(angles)]


def hull_traces_3d(pts, name, i, alpha=0.35, showlegend=True, edge_width=2.0):
    color = T10[i % len(T10)]
    try:
        hull = ConvexHull(pts)
    except Exception:
        loop_pts = order_points_on_plane(pts)
        loop = np.vstack([loop_pts, loop_pts[:1]])
        line = go.Scatter3d(x=loop[:, 0], y=loop[:, 1], z=loop[:, 2], mode="lines",
                             line=dict(color=color, width=edge_width * 2), name=name,
                             showlegend=showlegend, hovertemplate=f"{name}<extra></extra>")
        empty = go.Scatter3d(x=[], y=[], z=[], mode="lines", showlegend=False, hoverinfo="skip")
        return line, empty

    mesh = go.Mesh3d(x=pts[:, 0], y=pts[:, 1], z=pts[:, 2],
                      i=hull.simplices[:, 0], j=hull.simplices[:, 1], k=hull.simplices[:, 2],
                      color=color, opacity=alpha, flatshading=True,
                      name=name, showlegend=showlegend, hovertemplate=f"{name}<extra></extra>")
    ex, ey, ez, seen = [], [], [], set()
    for simplex in hull.simplices:
        for a, b in ((simplex[0], simplex[1]), (simplex[1], simplex[2]), (simplex[2], simplex[0])):
            key = tuple(sorted((int(a), int(b))))
            if key in seen:
                continue
            seen.add(key)
            ex += [pts[a, 0], pts[b, 0], None]
            ey += [pts[a, 1], pts[b, 1], None]
            ez += [pts[a, 2], pts[b, 2], None]
    edges = go.Scatter3d(x=ex, y=ey, z=ez, mode="lines", line=dict(color="black", width=edge_width),
                          showlegend=False, hoverinfo="skip")
    return mesh, edges


def curve_trace_2d(xy, name, color="black", dash=None, width=2.0):
    return go.Scatter(x=xy[:, 0], y=xy[:, 1], mode="lines", name=name,
                       line=dict(color=color, width=width, dash=dash))


def curve_trace_3d(xyz, name, color="black", width=4.0):
    return go.Scatter3d(x=xyz[:, 0], y=xyz[:, 1], z=xyz[:, 2], mode="lines", name=name,
                         line=dict(color=color, width=width))


def marker_trace_2d(xy, name, color="black", symbol="circle", size=9, showlegend=True):
    xy = np.atleast_2d(xy)
    return go.Scatter(x=xy[:, 0], y=xy[:, 1], mode="markers", name=name, showlegend=showlegend,
                       marker=dict(color=color, size=size, symbol=symbol, line=dict(color="black", width=1)),
                       hovertemplate=f"{name}<extra></extra>")


def square_xy(fig, row=None, col=None, xaxis="x", yaxis="y"):
    kwargs = dict(row=row, col=col) if row is not None else {}
    fig.update_yaxes(scaleanchor=xaxis, scaleratio=1, **kwargs)


def footprint_tube_xy(xy, radius, n_cap=24):
    """Boundary of the Minkowski sum of a 2D polyline with a disk of `radius` -- the
    robot's swept footprint band: offset curves running along both sides of the
    path, closed off with a rounded semicircular cap at each end."""
    tangents = np.empty_like(xy)
    tangents[1:-1] = xy[2:] - xy[:-2]
    tangents[0] = xy[1] - xy[0]
    tangents[-1] = xy[-1] - xy[-2]
    norms = np.linalg.norm(tangents, axis=1, keepdims=True)
    tangents = tangents / np.where(norms == 0, 1.0, norms)
    normals = np.column_stack([-tangents[:, 1], tangents[:, 0]])

    left = xy + radius * normals
    right = xy - radius * normals

    angle_left0 = np.arctan2(normals[0, 1], normals[0, 0])
    start_angles = np.linspace(angle_left0 + np.pi, angle_left0, n_cap)
    cap_start = xy[0] + radius * np.column_stack([np.cos(start_angles), np.sin(start_angles)])

    angle_left_end = np.arctan2(normals[-1, 1], normals[-1, 0])
    end_angles = np.linspace(angle_left_end, angle_left_end - np.pi, n_cap)
    cap_end = xy[-1] + radius * np.column_stack([np.cos(end_angles), np.sin(end_angles)])

    return np.vstack([left, cap_end, right[::-1], cap_start])
```

## Part 1 -- a straight corridor: `order=4` reproduces `order=2` exactly

Purely axis-aligned motion (no diagonal component), so this doesn't yet exercise
the velocity-model difference between the two orders (see Part 2's note) -- it
only confirms the Bezier machinery doesn't invent any slack or extra cost when a
straight line is already optimal.

```python
def build_straight_corridor(order, vlimit=1.0, tmax=20.0):
    """Two collinear rooms sharing a wall at x=2 -- the straight-line case where a
    cubic Bezier segment has no extra shape freedom to exploit; F1-F8's machinery
    should reproduce the order=2 answer bit-for-bit in cost."""
    g = STGCS([rect(0, 0, 4, 4)], t0=0.0, tmax=tmax, vlimit=vlimit, order=order)
    A = make_hpolytope(rect(0, 0, 2, 2))
    B = make_hpolytope(rect(2, 0, 4, 2))
    g.add_vertex(time_extruded(A, 0.0, tmax), Interval(0.0, tmax), name="A")
    g.add_vertex(time_extruded(B, 0.0, tmax), Interval(0.0, tmax), name="B")
    g.add_edges_bidir("A", "B")
    return g


g2 = build_straight_corridor(order=2)
sol2, rt2, status2 = run_search(g2, [0.5, 1.0], [3.5, 1.0])

g4 = build_straight_corridor(order=4)
sol4, rt4, status4 = run_search(g4, [0.5, 1.0], [3.5, 1.0])

print(f"order=2: status={status2.value:8s} duration={sol2.duration:.6f}  (STTrajectory)")
print(f"order=4: status={status4.value:8s} duration={sol4.duration:.6f}  (STSplineTrajectory)")
assert isinstance(sol2, STTrajectory) and isinstance(sol4, STSplineTrajectory)
assert abs(sol2.duration - sol4.duration) < 1e-4, "order=2 vs order=4 duration should match on a collinear case"
```

## Part 2 -- a genuine bend: an L-shaped corridor

Two rooms, `A = [0,3]x[0,1]` (a horizontal arm) and `B = [0,1]x[0,3]` (a vertical
arm), overlapping only in the `[0,1]x[0,1]` corner square. Start and goal sit at the
far ends of each arm, so the straight-line chord between them exits the union of
`A` and `B` entirely -- any feasible trajectory *must* bend through the shared
corner. Because `order>2` containment applies to a whole segment's control-point
hull (not just sampled points along it), the bend shows up as a real kink at the
`A`/`B` transition, exactly where `order=2`'s polyline also kinks -- the Bezier
curve can only round that corner, not avoid the kink altogether, since each
segment is private to one GCS vertex (F2/F3).

```python
def build_bent_corridor(order, vlimit=1.0, tmax=20.0):
    g = STGCS([rect(0, 0, 6, 6)], t0=0.0, tmax=tmax, vlimit=vlimit, order=order)
    A = make_hpolytope(rect(0.0, 0.0, 3.0, 1.0))
    B = make_hpolytope(rect(0.0, 0.0, 1.0, 3.0))
    g.add_vertex(time_extruded(A, 0.0, tmax), Interval(0.0, tmax), name="A")
    g.add_vertex(time_extruded(B, 0.0, tmax), Interval(0.0, tmax), name="B")
    g.add_edges_bidir("A", "B")
    return g


START, GOAL = [2.5, 0.5], [0.5, 2.5]

gb2 = build_bent_corridor(order=2)
solb2, _, statusb2 = run_search(gb2, START, GOAL)

gb4 = build_bent_corridor(order=4)
solb4, _, statusb4 = run_search(gb4, START, GOAL)

print(f"order=2 (polyline): status={statusb2.value:8s} duration={solb2.duration:.6f}")
print(f"order=4 (Bezier):   status={statusb4.value:8s} duration={solb4.duration:.6f}")
```

**Why `order=4`'s duration is *longer* here, not shorter, even though the curve
looks smoother:** this corridor requires genuinely diagonal motion, and the two
orders use deliberately different velocity models
(`AGENT/search-based-bezier-implementation-plan.md` SS3.2, resolved). `order=2`'s
existing constraint is an **L-infinity box** (independent per-axis bounds) --
moving at `(vx, vy) = (1, 1)` is legal, since each axis individually sits at the
limit, even though its Euclidean speed is `sqrt(2)`. `order=4` uses the **L2/SOC**
form F8 actually proves for a Bezier segment's derivative-control certificate --
the same diagonal move would violate `||v|| <= vlimit`, so the solver must slow
down on the diagonal, costing more time for the same displacement. Part 1 didn't
show this because axis-aligned motion makes the two velocity sets coincide;
Part 2's genuine diagonal bend is exactly where they diverge. This is the direct,
intended consequence of choosing the isotropic constraint for `order>2` -- not a
regression relative to `order=2`.

```python
fig = make_subplots(rows=1, cols=2, specs=[[{"type": "xy"}, {"type": "scene"}]],
                     subplot_titles=("top-down (x, y)", "space-time (rotate freely)"))

fig.add_trace(polygon_trace(rect(0.0, 0.0, 3.0, 1.0), "region A", 4, alpha=0.35), row=1, col=1)
fig.add_trace(polygon_trace(rect(0.0, 0.0, 1.0, 3.0), "region B", 0, alpha=0.35), row=1, col=1)

_, pts2 = sample_curve(solb2)
_, pts4 = sample_curve(solb4)
fig.add_trace(curve_trace_2d(pts2, "order=2 (polyline)", color="crimson", dash="dash", width=2.5), row=1, col=1)
fig.add_trace(curve_trace_2d(pts4, "order=4 (Bezier)", color="navy", width=3), row=1, col=1)
fig.add_trace(marker_trace_2d(START, "start", color="black"), row=1, col=1)
fig.add_trace(marker_trace_2d(GOAL, "goal", color="black", symbol="x"), row=1, col=1)

# space-time companion view: same two curves as (x, y, t) space-time paths, plus
# each region time-extruded into a translucent box for context (extruded only up
# to just past the longer trajectory's arrival time, not the corridor's full
# tmax=20, so the box doesn't dwarf the ~3s of actual motion).
t_extrude = 1.1 * max(solb2.duration, solb4.duration)


def box_pts(corners_xy, t0, t1):
    n = len(corners_xy)
    return np.vstack([np.column_stack([corners_xy, np.full(n, t0)]),
                       np.column_stack([corners_xy, np.full(n, t1)])])


box_a = box_pts(rect(0.0, 0.0, 3.0, 1.0), 0.0, t_extrude)
box_b = box_pts(rect(0.0, 0.0, 1.0, 3.0), 0.0, t_extrude)
mesh_a, edges_a = hull_traces_3d(box_a, "region A (space-time)", 4, alpha=0.06, showlegend=False)
mesh_b, edges_b = hull_traces_3d(box_b, "region B (space-time)", 0, alpha=0.06, showlegend=False)
for tr in (mesh_a, edges_a, mesh_b, edges_b):
    fig.add_trace(tr, row=1, col=2)

fig.add_trace(curve_trace_3d(pts2, "order=2 (polyline)", color="crimson", width=4), row=1, col=2)
fig.add_trace(curve_trace_3d(pts4, "order=4 (Bezier)", color="navy", width=5), row=1, col=2)
fig.add_trace(go.Scatter3d(x=[START[0]], y=[START[1]], z=[0.0], mode="markers",
                            marker=dict(size=5, color="black"), name="start", showlegend=False), row=1, col=2)

square_xy(fig, row=1, col=1)
fig.update_scenes(xaxis_title="x", yaxis_title="y", zaxis_title="t")
fig.update_layout(title="L-corridor: order=2 polyline vs order=4 Bezier", width=1050, height=560)
fig.show()
```

## Part 3 -- verifying the theory: F1 (hull), F3 (C0/C1 continuity), F8 (velocity)

Sample the `order=4` curve densely and check three of `AGENT.md`'s certificates
directly against this solved trajectory (the same checks as `AGENT.md` T1/T2(b)/T3):
every control point's segment boundary matches its neighbor (C0), the tangent
matches across every *interior* join -- i.e. every join except the one touching the
`source` vertex, which has no defined internal velocity and is intentionally C0-only
(F6) -- and the sampled speed never exceeds `vlimit`.

```python
for i in range(solb4.size - 1):
    cp_tail, cp_head = solb4.control_points(i), solb4.control_points(i + 1)
    c0_gap = np.linalg.norm(cp_tail[-1] - cp_head[0])
    is_boundary = solb4.vertex_path[i] == "source" or solb4.vertex_path[i + 1] == "source"
    if is_boundary:
        print(f"join {i} ({solb4.vertex_path[i]:>6s} -> {solb4.vertex_path[i+1]:<6s}): "
              f"C0 gap={c0_gap:.2e}  (boundary edge, C1 not required -- F6)")
        continue
    tail_tan, head_tan = cp_tail[-1] - cp_tail[-2], cp_head[1] - cp_head[0]
    c1_gap = np.linalg.norm(tail_tan - head_tan)
    print(f"join {i} ({solb4.vertex_path[i]:>6s} -> {solb4.vertex_path[i+1]:<6s}): "
          f"C0 gap={c0_gap:.2e}  C1 gap={c1_gap:.2e}")

ts, pts = sample_curve(solb4, n=300)
# pts includes the time coordinate as its last column (lerp returns the full
# joint (space, time) point) -- use only the spatial columns for a speed check,
# or the redundant time-vs-time term inflates the norm.
speed = np.linalg.norm(np.diff(pts[:, :-1], axis=0), axis=1) / np.diff(ts)
print(f"\nmax sampled speed = {speed.max():.6f}  (vlimit = {gb4.vlimit})")
assert speed.max() <= gb4.vlimit * 1.05, "velocity certificate (F8) violated"

# hull certificate (F1): every sampled point of segment i lies in conv(control_points(i))
for i in range(solb4.size):
    cps = solb4.control_points(i)
    lo, hi = cps.min(axis=0) - 1e-9, cps.max(axis=0) + 1e-9
    seg_ts = np.linspace(cps[0, -1], cps[-1, -1], 60)
    seg_pts = np.array([solb4.lerp(t) for t in seg_ts])
    assert np.all(seg_pts >= lo) and np.all(seg_pts <= hi), f"segment {i} hull certificate failed"
print("\nF1 hull-containment (bounding-box necessary condition) holds for every segment.")
```

## Part 4 -- space-time view: control points, hull, and the curve inside it (F1)

`AGENT.md` F4 requires time to be a coordinate of each joint control point, not an
annotation -- plot one segment's 4 joint control points `(x, y, t)`, the convex hull
they certify the curve lies in, and the curve itself, all in the same 3D space-time
axes.

```python
seg_idx = next(i for i, name in enumerate(solb4.vertex_path) if name not in ("source",))
cps = solb4.control_points(seg_idx)
seg_ts = np.linspace(cps[0, -1], cps[-1, -1], 100)
seg_pts = np.array([solb4.lerp(t) for t in seg_ts])

mesh, edges = hull_traces_3d(cps, "control-point hull", 0, alpha=0.25)
fig = go.Figure([mesh, edges])
fig.add_trace(go.Scatter3d(x=cps[:, 0], y=cps[:, 1], z=cps[:, 2], mode="markers+lines",
                            line=dict(color="black", width=2, dash="dash"),
                            marker=dict(size=5, color="black"), name="control polygon"))
fig.add_trace(curve_trace_3d(seg_pts, "curve", color="crimson", width=5))
fig.update_layout(
    title=f"Segment {seg_idx} ({solb4.vertex_path[seg_idx]}): control-point hull contains the curve (F1)",
    scene=dict(xaxis_title="x", yaxis_title="y", zaxis_title="t"),
    width=650, height=600,
)
fig.show()
```

## Part 5 -- reservation: Minkowski-inflated hull, then `reserve_spline`

Per `AGENT.md` SS3's reservation pipeline: inflate a segment's control-point hull by
the robot's footprint (F7), take the inflated hull's own facets as decision
boundaries, and slice the region graph against them -- `stgcs.ecd.reserve_spline`
does this once per segment across the whole trajectory, reusing the exact same
`slice`/`update_edge` graph machinery the `order=2` path already had.

```python
from stgcs.hulls import inflate_hull, minkowski_sum_vertices
from stgcs.ecd import parallelotope_verts_offset

safe_radius = 0.15
footprint = safe_radius * parallelotope_verts_offset(gb4.dimension + 1)
inflated = inflate_hull(cps, footprint)
raw_sum = minkowski_sum_vertices(cps, footprint)

mesh_raw, edges_raw = hull_traces_3d(cps, "raw hull", 0, alpha=0.15)
mesh_inf, edges_inf = hull_traces_3d(raw_sum, "inflated hull (F7)", 3, alpha=0.12)

fig = go.Figure([mesh_raw, edges_raw, mesh_inf, edges_inf])
fig.add_trace(curve_trace_3d(seg_pts, "curve", color="black", width=4))
fig.update_layout(
    title="Raw control-point hull (blue) vs Minkowski-inflated reservation hull (red), F7",
    scene=dict(xaxis_title="x", yaxis_title="y", zaxis_title="t"),
    width=650, height=600,
)
fig.show()

print(f"raw hull facets:      {ConvexHull(cps).simplices.shape[0]}")
print(f"inflated hull H-rep rows (after ReduceInequalities): {inflated.A().shape[0]}")
```

```python
n_before = gb4.G.number_of_nodes()
gb4_reserved = ecd.reserve_spline(gb4, solb4, safe_radius=safe_radius)
n_after = gb4_reserved.G.number_of_nodes()
print(f"graph fragmentation from one reservation: {n_before} vertices -> {n_after} vertices")
```

## Part 6 -- a second robot sees the fragmented graph

A second robot planning *after* the reservation is permanently restricted by
robot 1's swept-and-inflated corridor (this phase is fixed-priority, single-shot
per robot -- no rolling horizon, no re-ordering, per `AGENT.md` SS1). An offset
start/goal in the same two rooms is unaffected and solves immediately; the exact
same start/goal as robot 1 fails, because `reserve_spline`'s default
`x0_staying=True`/`xt_staying=True` parks robot 1 at its start and goal points for
all time outside its transit window -- the same convention `ecd.reserve()` already
used for `order=2`.

## Part 8 -- scalability stress test: 10 agents planning sequentially from empty space

Everything above reserves trajectories into a graph that already had named regions
(A/B) carved out of it by hand. This section starts from **one single convex
region -- the whole planning square, completely empty** -- and lets fixed-priority
planning alone do all the fragmenting: exactly `AGENT.md` SS1's phase-1 loop, run
end to end through the actual search-based planner (`SearchPlanner`, not the
`MICPPlanner` relaxation+rounding path Part 7 cross-checked). Agent `i` solves its
full-mission-horizon `order=4` program on whatever graph agent `i-1` left behind,
then permanently reserves its own solved trajectory (`ecd.reserve_spline`) into
that graph before agent `i+1` plans. Ten agents, fixed priority order, all released
at `t_start=0` (the maximally-contested case: every later agent must route around
every earlier agent's *entire* space-time reservation, not one offset in time),
independently randomized start/goal pairs per agent.

**This experiment is explicitly allowed to fail.** Per
`AGENT/search-based-bezier-implementation-plan.md` SS3.6 and the sibling
`notebooks/ecd_bezier_visualization.ipynb`'s own Part 9-12 investigation (run
against the whole-graph relaxation solver rather than this notebook's search
planner), repeated `reserve_spline` calls fragment the graph and
`add_edges_bidir`'s reconnection cost compounds with depth -- there is no
guarantee agent 9 or 10 solves in reasonable time, or at all, starting from a
single fully-empty region. Every step below is wrapped so one agent's search or
reservation failure/exception is recorded and the loop moves on, rather than
halting the other nine -- a mid-run failure is itself a valid, informative
result under `AGENT.md` SS1's fixed-priority semantics (infeasibility under a
fixed order is a planning failure, not a trigger to search over orderings).

```python
def build_empty_space(order, W=10.0, vlimit=1.0, tmax=60.0):
    """One GCS vertex, no internal structure -- the entire WxW square is free
    space, exactly the 'single vertex/convex hull containing all free space'
    starting point this stress test wants."""
    g = STGCS([rect(0.0, 0.0, W, W)], t0=0.0, tmax=tmax, vlimit=vlimit, order=order)
    free = make_hpolytope(rect(0.0, 0.0, W, W))
    g.add_vertex(time_extruded(free, 0.0, tmax), Interval(0.0, tmax), name="free")
    return g


def sample_start_goal(rng, W, margin, min_sep, max_tries=200):
    """Randomized start/goal within the square, kept off the walls (`margin`,
    room for the footprint's Minkowski inflation) and apart from each other
    (`min_sep`) so every agent actually has to cross the shared space."""
    for _ in range(max_tries):
        s = rng.uniform(margin, W - margin, size=2)
        g_pt = rng.uniform(margin, W - margin, size=2)
        if np.linalg.norm(g_pt - s) >= min_sep:
            return s, g_pt
    return s, g_pt  # fall back to the last draw if min_sep proved too strict


N_AGENTS = 10
STRESS_W = 10.0
STRESS_VLIMIT = 1.0
STRESS_TMAX = 60.0
STRESS_SAFE_RADIUS = 0.2
STRESS_SEED = 171

rng = np.random.default_rng(STRESS_SEED)
g_stress = build_empty_space(order=4, W=STRESS_W, vlimit=STRESS_VLIMIT, tmax=STRESS_TMAX)
print(f"initial graph: {g_stress.G.number_of_nodes()} vertex, {g_stress.G.number_of_edges()} edges "
      f"(the single {STRESS_W}x{STRESS_W} free-space region)")
```

```python
records = []
agent_solutions = []  # (agent index, start, goal, STSplineTrajectory) for every agent that solved
graph_snapshots = []  # g_stress.copy() taken right before each agent i searches (Part 9 needs the
                      # exact fragmented graph agent i saw, not the final post-loop g_stress)

t_wall0 = time.perf_counter()
for i in range(N_AGENTS):
    start, goal = sample_start_goal(rng, STRESS_W, margin=STRESS_SAFE_RADIUS + 0.5, min_sep=STRESS_W / 3)
    n_v_before, n_e_before = g_stress.G.number_of_nodes(), g_stress.G.number_of_edges()
    graph_snapshots.append(g_stress.copy())

    t0 = time.perf_counter()
    try:
        sol, _, status = run_search(g_stress, start, goal, t_start=0.0, vlimit=STRESS_VLIMIT)
        search_error = None
    except Exception as exc:
        sol, status, search_error = None, None, repr(exc)
    search_time_s = time.perf_counter() - t0

    rec = dict(agent=i, start=start.copy(), goal=goal.copy(),
               status=(status.value if status is not None else "error"),
               search_time_s=search_time_s, search_error=search_error,
               n_vertices_before=n_v_before, n_edges_before=n_e_before,
               duration=None, reserve_time_s=None, reserve_error=None)

    if status == STPlanStatus.SUCCESS:
        rec["duration"] = sol.duration
        agent_solutions.append((i, start, goal, sol))
        t1 = time.perf_counter()
        try:
            g_stress = ecd.reserve_spline(g_stress, sol, safe_radius=STRESS_SAFE_RADIUS, slice_method="fixed")
        except Exception as exc:
            rec["reserve_error"] = repr(exc)
        rec["reserve_time_s"] = time.perf_counter() - t1

    rec["n_vertices_after"] = g_stress.G.number_of_nodes()
    rec["n_edges_after"] = g_stress.G.number_of_edges()
    records.append(rec)

    flag = ""
    if rec["search_error"]:
        flag = f"  ERR(search)={rec['search_error']}"
    elif rec["reserve_error"]:
        flag = f"  ERR(reserve)={rec['reserve_error']}"
    print(f"agent {i:2d}: status={rec['status']:8s}  search={search_time_s:7.3f}s"
          f"  reserve={(rec['reserve_time_s'] if rec['reserve_time_s'] is not None else float('nan')):7.3f}s"
          f"  vertices {rec['n_vertices_before']:5d} -> {rec['n_vertices_after']:5d}"
          f"  edges {rec['n_edges_before']:6d} -> {rec['n_edges_after']:6d}{flag}")

total_wall_s = time.perf_counter() - t_wall0
n_success = sum(1 for r in records if r["status"] == "success")
print(f"\n{n_success}/{N_AGENTS} agents solved successfully; total wall time {total_wall_s:.1f}s")
```

### Verifying C1 continuity across all 10 agents

The same per-join check Part 3 ran for the single L-corridor trajectory (`AGENT.md`
F3/T2(b)): at every join between consecutive GCS-vertex segments within one agent's
path, the C0 gap (tail control point vs. head control point) should be zero to
solver tolerance, and the C1 gap (tail tangent vs. head tangent, raw control-point
differences per F3) should also vanish -- except at the join touching the
`source` vertex, which has no defined internal velocity and is intentionally
C0-only (F6). Run once per solved agent, not just the earlier hand-built example.

```python
c1_summary = []
for idx, start, goal, sol in agent_solutions:
    max_c0, max_c1, n_interior_joins = 0.0, 0.0, 0
    for i in range(sol.size - 1):
        cp_tail, cp_head = sol.control_points(i), sol.control_points(i + 1)
        c0_gap = np.linalg.norm(cp_tail[-1] - cp_head[0])
        max_c0 = max(max_c0, c0_gap)

        is_boundary = sol.vertex_path[i] == "source" or sol.vertex_path[i + 1] == "source"
        if is_boundary:
            continue
        tail_tan, head_tan = cp_tail[-1] - cp_tail[-2], cp_head[1] - cp_head[0]
        c1_gap = np.linalg.norm(tail_tan - head_tan)
        max_c1 = max(max_c1, c1_gap)
        n_interior_joins += 1

    c1_summary.append((idx, sol.size, n_interior_joins, max_c0, max_c1))
    print(f"agent {idx:2d}: {sol.size:3d} segments, {n_interior_joins:3d} interior joins  "
          f"max C0 gap={max_c0:.2e}  max C1 gap={max_c1:.2e}")

tol = 1e-6
all_c0_ok = all(c0 < tol for *_, c0, _ in c1_summary)
all_c1_ok = all(c1 < tol for *_, c1 in c1_summary)
print(f"\nAll {len(c1_summary)} agents' trajectories: C0-continuous={all_c0_ok}, "
      f"C1-continuous={all_c1_ok}  (tol={tol:.0e})")
assert all_c0_ok and all_c1_ok, "C0/C1 continuity certificate (F3) failed for at least one agent's trajectory"
```

**Reading the results (this run):** all 10 agents solved (10/10, ~130s total wall
time), but search time and reservation time both grow sharply with agent index --
search goes from 0.001s (agent 0, into the untouched 1-vertex graph) to 41.7s
(agent 9, into the 3438-vertex/47403-edge graph agent 8 left behind), and
reservation cost from 0.036s to 4.1s over the same span. The graph itself grows
from 1 vertex/0 edges to 4248 vertices/58663 edges -- edges outpacing vertices by
more than 10x, consistent with `add_edges_bidir`'s reconnection cost (an
`IntersectsWith` check per newly-adjacent-piece candidate, AABB-gated) being the
dominant driver, not the ECD splitting itself. This matches
`AGENT/search-based-bezier-implementation-plan.md` SS3.6 and
`ecd_bezier_visualization.ipynb`'s own Part 9-12 investigation (there, against the
whole-graph relaxation solver; here, reproduced end to end through the actual
search-based planner): reservation fragmentation is real and compounds with depth,
representation-independent, and "needs its own accounting," not a regression to
fix. All 10 succeeding is itself a run-dependent outcome, not a guarantee -- a
different `STRESS_SEED`, a smaller `STRESS_W`, or simply going past 10 agents could
just as easily hit a timeout or an exception partway through, which is exactly why
every step above is wrapped rather than assumed to succeed: per this section's own
framing, a mid-run failure would itself be a valid, informative result under
`AGENT.md` SS1's fixed-priority semantics, not something to paper over by retrying
with a different order.

## Part 9 -- visualizing fragmentation: why a "free" region still kinks

Part 8's loop overwrites `g_stress` in place each iteration, so there is no
record of what the graph looked like when any *specific* agent actually
searched it -- only the final, fully-fragmented graph survives. `graph_snapshots`
(a `.copy()` of `g_stress` taken right before each agent's search, added to the
loop above) makes it possible to go back and look at the exact graph one
heavily-detoured agent saw.

Each surviving cell is drawn as its spatial axis-aligned bounding box
(`STVertex.space_itvls` -- the same bound `add_edges_bidir`'s `AABB` check
already uses internally, not a re-derivation). A bounding box is a looser
approximation of a cell's true polytope shape, but the point here isn't
precision -- it's to make the *mosaic* visible: pick the agent whose solved path
threads through the most distinct cells, overlay that path and the specific
cells it was routed through against the full field of surviving fragments, and
compare both to the direct straight-line chord the agent would have taken had
the graph not been sliced up by earlier agents' reservations.


```python
# pick the agent whose solved path threads through the most distinct GCS-vertex
# cells -- the clearest case of "kinked despite free space" from the diagnosis
# above.
def n_real_cells(sol):
    return sum(1 for name in sol.vertex_path if name != "source")


frag_idx, frag_start, frag_goal, frag_sol = max(agent_solutions, key=lambda rec: n_real_cells(rec[3]))
frag_snapshot = graph_snapshots[frag_idx]
print(f"agent {frag_idx}: solved path threads through {n_real_cells(frag_sol)} cells, "
      f"graph had {frag_snapshot.G.number_of_nodes()} vertices / "
      f"{frag_snapshot.G.number_of_edges()} edges when it planned")

path_cells = [name for name in frag_sol.vertex_path if name != "source"]
all_cells = list(frag_snapshot.G.nodes)


def cell_rect(snapshot, name):
    v = snapshot.get_vertex(name)
    lo = [itvl.start for itvl in v.space_itvls]
    hi = [itvl.end for itvl in v.space_itvls]
    return rect(lo[0], lo[1], hi[0], hi[1])


# every surviving cell, drawn as thin outlines only -- there can be thousands, so
# this is one multi-polygon Scatter trace (None-separated loops), not one trace
# per cell, to keep rendering fast.
xs, ys = [], []
for name in all_cells:
    loop = cell_rect(frag_snapshot, name)
    loop = np.vstack([loop, loop[:1]])
    xs += list(loop[:, 0]) + [None]
    ys += list(loop[:, 1]) + [None]
outline_trace = go.Scatter(x=xs, y=ys, mode="lines", line=dict(color="lightgray", width=0.6),
                            name=f"{len(all_cells)} fragmented cells", hoverinfo="skip")

fig = go.Figure([outline_trace])
for k, name in enumerate(path_cells):
    fig.add_trace(polygon_trace(cell_rect(frag_snapshot, name), f"cell {name}", k, alpha=0.5, showlegend=False))

_, pts_frag = sample_curve(frag_sol)
straight = np.array([frag_start, frag_goal])
fig.add_trace(curve_trace_2d(pts_frag[:, :2], f"agent {frag_idx} solved path",
                              color="black", width=3))
fig.add_trace(go.Scatter(x=straight[:, 0], y=straight[:, 1], mode="lines",
                          line=dict(color="crimson", width=2, dash="dot"),
                          name="direct chord"))
fig.add_trace(marker_trace_2d(frag_start, f"agent {frag_idx} start", color="black"))
fig.add_trace(marker_trace_2d(frag_goal, f"agent {frag_idx} goal", color="black", symbol="x"))

fig.update_layout(
    title=f"Agent {frag_idx}: {len(path_cells)} threaded cells out of {len(all_cells)} fragments",
    xaxis_title="x", yaxis_title="y", width=800, height=850,
    legend=dict(orientation="h", yanchor="top", y=-0.08, xanchor="center", x=0.5),
)
square_xy(fig)
fig.show()
```
