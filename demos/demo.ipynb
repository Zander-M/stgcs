# MRMP Demo Instances

Run the YAML-defined demo instances through native Python calls into `MRMPDemo`.


```python
from pathlib import Path
import sys

root = Path.cwd().resolve()
if root.name == 'demos':
    root = root.parent
if str(root) not in sys.path:
    sys.path.insert(0, str(root))

from demos.mrmp import MRMPDemo

output_root = root / 'demos/runs'
serve_viewer = True
trajopt_progress = True
```

# ST-GCS Formation Control


```python
stgcs_formation_control_options = {
    'stage_timeout_secs': 300.0,
    'search_eps': 10.0,
    'coordination': 'pbs',
    'child_expansion_rule': 'num_conflicts',
}
stgcs_formation_control_task = MRMPDemo.load_task(
    root / 'demos/instances/stgcs_formation_control.yaml',
    solver_overrides=stgcs_formation_control_options,
)
stgcs_formation_control_paths = MRMPDemo.resolved_paths_from_values(
    stgcs_formation_control_task,
    output_dir=output_root,
)
stgcs_formation_control_result = MRMPDemo.run_pipeline(
    stgcs_formation_control_task,
    stgcs_formation_control_paths,
    trajopt_progress=trajopt_progress,
    serve=serve_viewer,
)
stgcs_formation_control_result
```
